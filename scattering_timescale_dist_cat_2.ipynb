```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../ondemand/data/scattering_times.csv")
```

```python
df = df.drop_duplicates(subset=['event_id'])
```

```python
df = df[df['scat_time'] > 0]
```

```python
df['log_scat'] = np.log(df['scat_time'])
```

```python
scatt_time = df['scat_time']
```

```python
fig, ax = plt.subplots()
sns.histplot(data=df, x='scat_time', binrange=[0, 200], binwidth=3, common_norm=True, stat="density",
            ax=ax, edgecolor='blue', element='bars', linewidth=1)
ax.grid(visible=True, color='grey', linestyle=':', linewidth=0.5, alpha=0.5)
# sns.kdeplot(data=df, x='scat_time', ax=ax, clip=[0, 200], bw_adjust=0.5, fill=True, cut=0)
# ax.set_xlim([-30, 200])
ax.set_xlabel(r"Scattering Time [ms] at $\nu_{ref}=400MHz$")
ax.minorticks_on()

for spine in ax.spines.values():
    spine.set_edgecolor('black', )
    spine.set_linewidth(1.5)

plt.savefig("scatt_distribution.jpg", dpi=600)
```

```python
fig, ax = plt.subplots()
sns.histplot(data=df, x='log_scat', binrange=[-3, 8], binwidth=0.2, common_norm=True, stat="density",
            ax=ax, edgecolor='blue', element='bars', linewidth=1)
ax.grid(visible=True, color='grey', linestyle=':', linewidth=0.5, alpha=0.5)
# sns.kdeplot(data=df, x='scat_time', ax=ax, clip=[0, 200], bw_adjust=0.5, fill=True, cut=0)
# ax.set_xlim([-30, 200])
ax.set_xlabel(r"Log Scattering Time")
ax.minorticks_on()

for spine in ax.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(1.5)

plt.savefig("log_scatt_time_dist.jpg", dpi=600)
```
